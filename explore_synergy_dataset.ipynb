```python
import os
import pandas as pd
from synergy_dataset import Dataset, iter_datasets
from dotenv import load_dotenv

load_dotenv()

# list the exact dataset names
print([d.name for d in iter_datasets()])

d = Dataset("Jeyaraman_2020")   
```

```python
# the default `abstract` var is SYNERGY+-only (empty here); classic data uses `abstract_original`
df = (
    d.to_frame(vars=["title", "abstract_original"])
     .rename(columns={"abstract_original": "abstract"})
)
df.insert(0, "id", range(len(df)))
df.head(3)
```

```python
# sanity checks: expect 1175 rows, 96 included, 2 empty abstracts
print(df.shape, df.label_included.value_counts().to_dict())
print("no abstract:", df.abstract.fillna("").str.strip().eq("").sum(
))
df.abstract.fillna("").str.split().str.len().describe()
```
